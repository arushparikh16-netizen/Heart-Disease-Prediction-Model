```python
df['Cholesterol']=df['Cholesterol'].replace(0,np.nan)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,f1_score,recall_score,precision_score,classification_report
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix
```

```python
x=df.drop(columns=['HeartDisease'])
y=df['HeartDisease']

xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42,stratify=y)
print(xtrain)
```

```python
cat_col=['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope']

imput_col=['Cholesterol']

num_col=['Age','RestingBP','FastingBS','MaxHR','Oldpeak']

pipeline1=OneHotEncoder()

pipeline2=StandardScaler()

pipleline3=Pipeline([
    ('simpleimput',SimpleImputer(strategy='median')),
    ('stnad',StandardScaler())
])

pre=ColumnTransformer([
    ('c1',pipeline1,cat_col),
    ('c2',pipeline2,num_col),
    ('c3',pipleline3,imput_col)
])

model=Pipeline([
    ('preprocessor',pre),
    ('lr',LogisticRegression(max_iter=2000,C=0.1))
])

model.fit(xtrain,ytrain)

ypred=model.predict(xtest)
print(ypred)
print(ytest)

# print(model.predict_proba(xtest))

print(confusion_matrix(ytest,ypred))
print("Accuracy of model is:",accuracy_score(ytest,ypred))
print("Recall score of model is:",recall_score(ytest,ypred))
print("Precision Score of model is:",precision_score(ytest,ypred))
print("F1 score of model is:",f1_score(ytest,ypred))

print(classification_report(ytest,ypred))
```
